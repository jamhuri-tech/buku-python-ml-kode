# 4 · `break`, `continue`, dan `for ... else`

**[COBA]** `continue` melewati sisa tubuh perulangan. `break` menghentikan perulangan terdekat.

```python
total = 0
for nilai in [10, -5, 20, -3, 30]:
    if nilai < 0:
        continue
    total += nilai
print("jumlah nilai tak negatif:", total)
```

**Definisi `cari`** · bagian 1 dari 2

```python
def cari(data, target):
    for nilai in data:
        if nilai == target:
            print(target, "ditemukan")
            break
    else:
        print(target, "tidak ditemukan")
```

**Jalankan contoh atau lanjutkan langkah sebelumnya** · bagian 2 dari 2

# 5 · Jebakan mengubah `list` saat menyusurinya

**[DISKUSI]** Bandingkan dua data berikut. Salah satunya kebetulan memberi hasil yang tampak benar, sedangkan yang lain melewatkan unsur. Sebelum menjalankan, prediksi hasil keduanya.

**Definisi `hapus_genap_sambil_menyusuri`** · bagian 1 dari 2

```python
def hapus_genap_sambil_menyusuri(angka):
    angka = angka.copy()
    for nilai in angka:
        if nilai % 2 == 0:
            angka.remove(nilai)
    return angka
```

**Jalankan contoh atau lanjutkan langkah sebelumnya** · bagian 2 dari 2

```python
for data in ([1, 2, 3, 4], [1, 2, 4, 5]):
    print(data, "->", hapus_genap_sambil_menyusuri(data))
```
