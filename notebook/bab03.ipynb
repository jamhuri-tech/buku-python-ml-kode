## 1. Percabangan bertingkat

````python
def kategori_pesanan(jumlah):
    if jumlah >= 100:
        return "borongan"
    elif jumlah >= 20:
        return "besar"
    elif jumlah > 0:
        return "eceran"
    else:
        return "tidak sah"

print(kategori_pesanan(150), kategori_pesanan(5))
````

## 10. Melewati nilai tertentu

## 11. Perulangan dengan else

````python
def cari(data, target):
    for v in data:
        if v == target:
            print(f"{target} ditemukan")
            break
    else:
        print(f"{target} tidak ada sama sekali")

cari([1, 2, 3], 2)
cari([1, 2, 3], 9)
````

Keluaran yang diharapkan:

```
2 ditemukan
9 tidak ada sama sekali
```

## 12. Yang tampak masuk akal

## 13. Cara yang benar

## 14. Dengan penanda

## 15. Dengan fungsi

````python
def cari(m, t):
    for i, baris in enumerate(m):
        for j, v in enumerate(baris):
            if v == t:
                return i, j
    return None

print(cari(matriks, 5))
print(cari(matriks, 99))
````

Keluaran yang diharapkan:

```
(1, 1)
None
```
