## 6. Membuat simbol

````python
x, y = sp.symbols("x y")

e = (x**2 - 1)/(x - 1)

sp.simplify(e)
e.subs(x, 2)
````

## 7. Turunan biasa dan parsial

````python
f = sp.sin(x)*sp.exp(-x**2)

sp.diff(f, x)
sp.simplify(sp.diff(f, x, 2))

g = x**2*y + sp.sin(x*y)
sp.diff(g, x)
sp.diff(g, x, y)
````

## 8. Gradien dan matriks Hesse

````python
v = sp.Matrix([x, y])
grad = sp.Matrix([sp.diff(g, s) for s in v])
hess = sp.hessian(g, v)
````

## 9. Tiga jenis integral

````python
sp.integrate(x**2*sp.sin(x), x)
sp.integrate(sp.sin(x), (x, 0, sp.pi))
sp.integrate(sp.exp(-x**2), (x, -sp.oo, sp.oo))
````

## 10. Limit dan deret Taylor

````python
sp.limit(sp.sin(x)/x, x, 0)
sp.limit((1 + 1/x)**x, x, sp.oo)
sp.series(sp.exp(x), x, 0, 5)
sp.series(sp.sin(x), x, 0, 8).removeO()
````

## 11. Menyelesaikan persamaan

````python
sp.solve(x**2 - 4, x)
sp.solve([x + y - 3, x - y - 1], [x, y])
sp.nsolve(sp.cos(x) - x, x, 1)
````

## 12. Operasi matriks

````python
A = sp.Matrix([[2, 1],
               [1, 3]])
b = sp.Matrix([5, 10])

A.det()
A.inv()
A.eigenvals()
A.solve(b).T
````

## 13. Matriks bersimbol

````python
M = sp.Matrix([[1, x],
               [x, 1]])
M.det()
sp.simplify(M.inv())
````

## 14. Persamaan orde satu

````python
t = sp.symbols("t")
y = sp.Function("y")

pdb = sp.Eq(y(t).diff(t), -2*y(t))

sp.dsolve(pdb, y(t))
sp.dsolve(pdb, y(t), ics={y(0): 5})
````

## 15. Persamaan orde dua dengan syarat awal

````python
pdb = sp.Eq(y(t).diff(t, 2) + 2*y(t).diff(t) + 5*y(t), 0)

sol = sp.dsolve(pdb, y(t),
                ics={y(0): 1,
                     y(t).diff(t).subs(t, 0): 0})
sp.simplify(sol)
````

## 16. Memeriksa jawaban

````python
s = sol.rhs
sp.simplify(s.diff(t, 2) + 2*s.diff(t) + 5*s)
````

## 17. Sistem dua persamaan

````python
f, g = sp.Function("f"), sp.Function("g")

sistem = [sp.Eq(f(t).diff(t), f(t) + 2*g(t)),
          sp.Eq(g(t).diff(t), 3*f(t) + 2*g(t))]

[sp.simplify(h) for h in sp.dsolve(sistem)]
````

## 18. Menjembatani keduanya

````python
import numpy as np

f = sp.sin(x)*sp.exp(-x**2/2)
fp = sp.diff(f, x)

fn = sp.lambdify(x, fp, "numpy")

xs = np.linspace(-3, 3, 5)
print(np.round(fn(xs), 6))
````

## 19. Penurunan mandiri oleh SymPy

````python
n, d = 8, 3
w0, w1, w2 = sp.symbols("w0 w1 w2")
w = sp.Matrix([w0, w1, w2])

X = sp.MatrixSymbol("X", n, d)
Y = sp.MatrixSymbol("y", n, 1)

r = sp.Matrix(X)*w - sp.Matrix(Y)
J = (r.T*r)[0, 0] / n

grad = sp.Matrix([sp.diff(J, v) for v in (w0, w1, w2)])
rumus = sp.Rational(2, n)*sp.Matrix(X).T*r

sp.simplify(grad - rumus).T
````
