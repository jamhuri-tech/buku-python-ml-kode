## 1. Fungsi pertama

````python
def omzet(jumlah, harga):
    return jumlah * harga

print(omzet(3, 85000))
````

Keluaran yang diharapkan:

```

```

## 2. Empat cara memanggil

````python
def transaksi(produk, jumlah, harga, diskon=0.0):
    return jumlah * harga * (1 - diskon)

print(transaksi("Nastar", 3, 85000))
print(transaksi(produk="Nastar", harga=85000, jumlah=3))
print(transaksi("Nastar", 3, harga=85000, diskon=0.1))
````

Keluaran yang diharapkan:

```
255000.0
255000.0
229500.0
```

## 3. Hasil yang mengejutkan

````python
def tambah(item, ke=[]):
    ke.append(item)
    return ke

print(tambah("a"), tambah("b"), tambah("c"))
````

Keluaran yang diharapkan:

```
['a', 'b', 'c'] ['a', 'b', 'c'] ['a', 'b', 'c']
```

## 4. Pola None

````python
def tambah(item, ke=None):
    if ke is None:
        ke = []
    ke.append(item)
    return ke

print(tambah("a"), tambah("b"), tambah("c"))
````

Keluaran yang diharapkan:

```
['a'] ['b'] ['c']
```

## 5. args dan kwargs

````python
def ringkas(*nilai, **opsi):
    return len(nilai), sum(nilai), opsi

print(ringkas(1, 2, 3, pembulatan=2, satuan="kg"))
````

Keluaran yang diharapkan:

```
(3, 6, {'pembulatan': 2, 'satuan': 'kg'})
```

## 6. Pembongkaran saat memanggil

````python
def f(a, b, c):
    return a + b + c

arg = [1, 2, 3]
kw = {"a": 1, "b": 2, "c": 3}

print(f(*arg), f(**kw))
````

Keluaran yang diharapkan:

```
6 6
```

## 7. Khusus posisi dan khusus kata kunci

````python
def skala(data, /, *, faktor=1.0):
    return [x * faktor for x in data]

print(skala([1, 2, 3], faktor=2))
skala(data=[1, 2, 3], faktor=2)
skala([1, 2, 3], 2)
````

Keluaran yang diharapkan:

```
[2, 4, 6]
TypeError: skala() got some positional-only arguments
           passed as keyword arguments: 'data'
TypeError: skala() takes 1 positional argument but
           2 were given
```

## 9. Ketiganya memberi jawaban sama

````python
fungsi = []
for i in range(3):
    fungsi.append(lambda: i)

print([f() for f in fungsi])
````

Keluaran yang diharapkan:

```
[2, 2, 2]
```

## 10. Menangkap nilai saat itu juga

````python
fungsi = []
for i in range(3):
    fungsi.append(lambda i=i: i)

print([f() for f in fungsi])
````

Keluaran yang diharapkan:

```
[0, 1, 2]
```

## 11. Fungsi sebagai argumen

````python
produk = [("Nastar", 85000),
          ("Brownies", 45000),
          ("Kastengel", 95000)]

menurut_harga = sorted(produk, key=lambda p: p[1])
menurut_nama = sorted(produk, key=lambda p: p[0])

print([p[0] for p in menurut_harga])
print([p[0] for p in menurut_nama])
````

Keluaran yang diharapkan:

```
['Brownies', 'Nastar', 'Kastengel']
['Brownies', 'Kastengel', 'Nastar']
```

## 12. Dokumentasi dan tipe

````python
def dengan_ppn(nilai: float, tarif: float = 0.11) -> float:
    """Mengembalikan nilai setelah ditambah PPN.

    nilai : jumlah sebelum pajak
    tarif : tarif pajak, bawaan 0.11
    """
    return nilai * (1 + tarif)

print(dengan_ppn.__doc__.splitlines()[0])
````

Keluaran yang diharapkan:

```
Mengembalikan nilai setelah ditambah PPN.
```

## 13. Berkas hitung.py

````python
"""Modul contoh: perhitungan penjualan."""
PPN = 0.11

def omzet(jumlah, harga):
    return jumlah * harga

def dengan_ppn(nilai):
    return nilai * (1 + PPN)

if __name__ == "__main__":
    print(dengan_ppn(omzet(3, 85000)))
````
