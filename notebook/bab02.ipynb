## 1. Dua label, satu objek

````python
a = [1, 2, 3]
b = a
print(id(a) == id(b))

b = [1, 2, 3]
print(id(a) == id(b), a == b)
````

Keluaran yang diharapkan:

```
True
False True
```

## 2. Enam tipe yang paling sering ditemui

````python
for v in [7, 7.0, True, "tujuh", None, 7+0j]:
    print(v, type(v).__name__)
````

Keluaran yang diharapkan:

```
7 int
7.0 float
True bool
tujuh str
None NoneType
(7+0j) complex
```

## 3. Hasil yang mengejutkan

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(format(0.1, ".20f"))
````

## 4. Galat yang menumpuk

````python
s = 0.0
for _ in range(10):
    s += 0.1

print(repr(s))
print(s == 1.0)
````

Keluaran yang diharapkan:

```
0.9999999999999999
False
```

## 5. Decimal untuk nilai uang

````python
from decimal import Decimal

print(Decimal("0.1") + Decimal("0.2"))
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))
````

Keluaran yang diharapkan:

```
0.3
True
```

## 6. Pembulatan bankir

## 11. Format dengan f-string

````python
nama, harga, jumlah = "Nastar", 85000, 3

print(f"{nama}: {jumlah} x {harga} = {jumlah*harga}")
print(f"{jumlah*harga:,}")
print(f"{harga/3:.2f}")
print(f"|{nama:>12}|{nama:<12}|{nama:^12}|")
print(f"{harga=}")
````

Keluaran yang diharapkan:

```
Nastar: 3 x 85000 = 255000
255,000
28333.33
|      Nastar|Nastar      |   Nastar   |
harga=85000
```
