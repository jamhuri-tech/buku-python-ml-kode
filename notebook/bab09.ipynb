## 1. Gaya pyplot

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [2, 4, 9])
plt.xlabel("x")
plt.title("gaya pyplot")
plt.show()
```

## 2. Gaya berorientasi objek

```python
fig, ax = plt.subplots()

ax.plot([1, 2, 3], [2, 4, 9])
ax.set_xlabel("x")
ax.set_title("gaya berorientasi objek")
```

## 3. Kode Gambar~gbr:anatomi-mpl

```python
import numpy as np

fig, ax = plt.subplots(figsize=(5.3, 3.2))

x = np.linspace(0, 10, 200)
ax.plot(x, np.sin(x), color="#1B3B6F", label="sin(x)")
ax.plot(x, np.cos(x), color="#B85C00", ls="--",
        label="cos(x)")

ax.set_xlabel("sumbu x")
ax.set_ylabel("sumbu y")
ax.set_title("judul Axes")
ax.set_ylim(-2.3, 1.9)
ax.legend(loc="lower left")
```

## 4. Empat gaya garis

```python
fig, ax = plt.subplots(figsize=(5.3, 2.9))
t = np.linspace(0, 4, 9)

ax.plot(t, t**0.5, ls="-", marker="o", ms=4,
        label="garis utuh + bulat")
ax.plot(t, t**0.5 * 0.8, ls="--", marker="s", ms=4,
        label="putus + kotak")
ax.plot(t, t**0.5 * 0.6, ls=":", marker="^", ms=5,
        label="titik + segitiga")
ax.plot(t, t**0.5 * 0.4, ls="-.", marker="d", ms=4,
        label="titik-garis + wajik")

ax.set_xlabel("t")
ax.set_ylabel("nilai")
ax.legend(fontsize=7.5, ncol=2)
```

## 5. Kisi dua kali dua

```python
fig, axs = plt.subplots(2, 2, figsize=(5.3, 3.6),
                        sharex="col")
xs = np.linspace(-3, 3, 200)

axs[0, 0].plot(xs, xs**2)
axs[0, 0].set_title("kuadrat")
axs[0, 1].plot(xs, np.tanh(xs))
axs[0, 1].set_title("tanh")
axs[1, 0].plot(xs, np.abs(xs))
axs[1, 0].set_title("nilai mutlak")
axs[1, 1].plot(xs, 1 / (1 + np.exp(-xs)))
axs[1, 1].set_title("sigmoid")

for ax in axs[1, :]:
    ax.set_xlabel("x")
```

## 6. Tiga jenis grafik dasar

```python
rng = np.random.default_rng(3)
fig, axs = plt.subplots(1, 3, figsize=(5.3, 2.2))

a = rng.normal(0, 1, 200)
b = 0.7 * a + rng.normal(0, 0.6, 200)
axs[0].scatter(a, b, s=8, alpha=0.6, edgecolor="none")
axs[0].set_title("scatter")

axs[1].hist(rng.normal(0, 1, 800), bins=25,
            edgecolor="white", linewidth=0.4)
axs[1].set_title("histogram")

axs[2].bar(["A", "B", "C", "D"], [23, 41, 17, 35],
           edgecolor="white")
axs[2].set_title("bar")
```

## 7. Membandingkan dua peta warna

```python
fig, axs = plt.subplots(1, 2, figsize=(5.3, 1.95))

for ax, cmap in zip(axs, ["jet", "viridis"]):
    im = ax.imshow(medan, cmap=cmap, origin="lower",
                   extent=[-3, 3, -2, 2], aspect="auto")
    ax.set_title(cmap)
    fig.colorbar(im, ax=ax, fraction=0.046)
```

## 8. Menyimpan untuk dua keperluan

```python
fig.savefig("gambar.pdf")              # vektor, untuk cetak
fig.savefig("gambar.png", dpi=300)     # raster, untuk layar
fig.savefig("rapat.pdf", bbox_inches="tight")
```

## 9. Riwayat konvergensi

```python
fig, ax = plt.subplots(figsize=(5.3, 2.9))

for eta in [0.01, 0.1, 0.5]:
    ax.semilogy(riwayat(eta), lw=1.6,
                label=f"$\\eta = {eta}$")

ax.axhline(26.366, color="0.5", ls="--", lw=0.9)
ax.set_xlabel("iterasi")
ax.set_ylabel("$J(w)$")
ax.legend()
```
