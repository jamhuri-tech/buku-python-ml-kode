## 2. Skrip pemeriksa lingkungan

```python
import sys

WAJIB = {"numpy": "2.0", "pandas": "3.0",
         "matplotlib": "3.8", "sklearn": "1.5",
         "scipy": "1.11"}

def versi_ok(ada, minimal):
    a = [int(x) for x in ada.split(".")[:2] if x.isdigit()]
    m = [int(x) for x in minimal.split(".")[:2]]
    return a >= m

print(f"Python  {sys.version.split()[0]}")
print(f"Lokasi  {sys.executable}")
aktif = sys.prefix != sys.base_prefix
status = "ya" if aktif else "TIDAK"
print(f"Lingkungan virtual aktif: {status}")

kurang = []
for nama, minimal in WAJIB.items():
    try:
        m = __import__(nama)
        v = getattr(m, "__version__", "?")
        tanda = "ok " if versi_ok(v, minimal) else "TUA"
        print(f"  [{tanda}] {nama:12s} {v}")
        if tanda == "TUA":
            kurang.append(nama)
    except ImportError:
        print(f"  [ -- ] {nama:12s} belum terpasang")
        kurang.append(nama)

if kurang:
    print("python -m pip install -U " + " ".join(
        "scikit-learn" if k == "sklearn" else k
        for k in kurang))
else:
    print("Lingkungan siap.")
```
