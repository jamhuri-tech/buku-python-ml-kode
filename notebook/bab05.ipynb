## 2. Dangkal versus dalam

````python
import copy

asal = [[1, 2], [3, 4]]
dangkal = asal.copy()
dangkal[0].append(99)
print(asal)

asal = [[1, 2], [3, 4]]
dalam = copy.deepcopy(asal)
dalam[0].append(99)
print(asal)
````

Keluaran yang diharapkan:

```
[[1, 2, 99], [3, 4]]
[[1, 2], [3, 4]]
```

## 7. Operasi himpunan

````python
a = {"toko", "online", "reseller"}
b = {"online", "marketplace"}

print(sorted(a | b))    # gabungan
print(sorted(a & b))    # irisan
print(sorted(a - b))    # selisih

kanal = ["toko", "online", "toko", "reseller", "online"]
print(sorted(set(kanal)))
````

Keluaran yang diharapkan:

```
['marketplace', 'online', 'reseller', 'toko']
['online']
['reseller', 'toko']
['online', 'reseller', 'toko']
```

## 8. Mengukur sendiri

````python
import timeit

for n in [1_000, 100_000, 1_000_000]:
    siap = (f"dl = list(range({n})); "
            f"ds = set(dl); cari = {n} - 1")
    tl = min(timeit.repeat("cari in dl", setup=siap,
                           repeat=5, number=100)) / 100
    ts = min(timeit.repeat("cari in ds", setup=siap,
                           repeat=5, number=100)) / 100
    print(n, round(tl*1e6, 2), round(ts*1e6, 3))
````

Keluaran yang diharapkan:

```
1000       5.43   0.020
100000   592.43   0.043
1000000 5982.78   0.073
```

## 13. Membongkar biaya sebuah list

````python
import sys

n = 1_000_000
lst = [float(i) for i in range(n)]

kerangka = sys.getsizeof(lst)
per_objek = sys.getsizeof(1.5)

print(kerangka / 1024**2)
print(per_objek)
print((kerangka + n*per_objek) / 1024**2)
print(n * 8 / 1024**2)
````

Keluaran yang diharapkan:

```
8.06
24
30.95
7.63
```
